```python
import matplotlib.pyplot as plt
from itertools import cycle
import numpy as np

# blue = "#466eff"
# green = "#64B478"
# red = "#EB423D"
# orange = "#FF883D"
# color_list = [blue, green, red, orange]
color_list = [
    "#466eff", "#66CCEE", "#64B478",
    "#88DD44", "#EB423D", "#77423D",
    "#BBBBBB", "#EEEE00", "#FF883D"
]
markers = cycle(["o", "s", "^", "D", "P", "X", "v", "*", "+"])
```

```python
# Ablation for activation function
params = [1e5, 1e6, 1e7, 1e8]
activation_list=['relu', 'tanh', 'elu', 'selu', 'gelu', 'leakyrelu', 'softpls', 'silu', 'none']
performance = [
    [86.54145, 85.40458, 77.19228, 64.18133], # ReLU
    [87.01748, 87.01748, np.NaN, 53.58038], # tanh
    [86.03512, 86.03512, 86.03512, 48.72715], # ELU
    [85.30403, 85.30403, 85.30403, 37.43315], # SELU
    [86.52305, 82.31578, 76.40119, 62.17372], # GELU
    [86.54523, 82.62015, 76.94374, 63.98057], # leakyrelu
    [86.71043, 82.31574, 74.66026, 57.68511], # softplus
    [86.68857, 82.14236, 75.62375, 60.98463], # silu
    [85.40458, 77.20049, 65.46317, 40.18023], # none
]

# Create the plot
fig, ax = plt.subplots(figsize=(5, 4))
for idx, perf in enumerate(performance):
	plt.plot(
		params, perf, label=activation_list[idx], 
		color=color_list[idx], linewidth=2, 
  		markersize=8, marker=next(markers)
	)

# Customize the plot
plt.title("PPBR_AZ")
plt.xlabel('Number of Parameters', fontsize=12)
plt.ylabel('MAE', fontsize=12)
plt.grid(True, alpha=0.3)
plt.axhline(y=6.454, color='black', linestyle='--', linewidth=2)
plt.legend(fontsize=11, loc='center left', bbox_to_anchor=(1.05, 0.5))
ax = plt.gca()
ax.set_xscale("log")
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, p: f'{x/1e6:.1f}M'))

plt.show()
```
