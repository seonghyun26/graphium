# Merge

```python
df_protein = pd.read_csv("./protein-drug.csv")
df_protein
```

```python
import pandas as pd

df_csv = df_csv.copy()
df_protein = df_protein.copy()

# 0) Normalize join keys (avoid int vs str mismatches)
df_csv["protein_id"] = df_csv["protein_id"].astype(str).str.strip()
df_protein["Target_ID"] = df_protein["Target_ID"].astype(str).str.strip()

# (Optional) normalize sequences too (sometimes have whitespace/newlines)
df_csv["sequence"] = df_csv["sequence"].astype(str).str.replace(r"\s+", "", regex=True)
df_protein["Target"] = df_protein["Target"].astype(str).str.replace(r"\s+", "", regex=True)

# 1) Diagnostics: is Target_ID -> Drug (SMILES) unique?
# Rename Drug -> SMILES in a temporary view for clarity
tmp = df_protein.rename(columns={"Drug": "SMILES"})

smiles_per_target = tmp.groupby("Target_ID")["SMILES"].nunique(dropna=True)
multi = smiles_per_target[smiles_per_target > 1]
if len(multi) > 0:
    print("WARNING: Some Target_ID map to multiple SMILES. Examples:")
    print(multi.head(10))

# 2) Build a deterministic mapping table Target_ID -> SMILES
# Rule: take first non-null SMILES per Target_ID (change rule if you prefer)
target_to_smiles = (
    tmp.dropna(subset=["SMILES"])
       .drop_duplicates(subset=["Target_ID"], keep="first")[["Target_ID", "SMILES"]]
)

# 3) Merge onto df_csv with validation (many proteins in df_csv -> one SMILES per target)
df_csv_smiles = df_csv.merge(
    target_to_smiles,
    left_on="protein_id",
    right_on="Target_ID",
    how="left",
    validate="m:1",
)

df_csv_smiles = df_csv_smiles.drop(columns=["Target_ID"])

# 4) (Optional) sanity check: sequence consistency for matched rows
seq_map = tmp.drop_duplicates(subset=["Target_ID"])[["Target_ID", "Target"]]
seq_check = df_csv.merge(seq_map, left_on="protein_id", right_on="Target_ID", how="left")
seq_mismatch_rate = (
    (seq_check["Target"].notna()) &
    (seq_check["sequence"] != seq_check["Target"])
).mean()
print(f"Sequence mismatch rate (where target sequence exists): {seq_mismatch_rate:.2%}")

# 5) Report coverage
missing_rate = df_csv_smiles["SMILES"].isna().mean()
print(f"Missing SMILES rate after merge: {missing_rate:.2%}")
```

```python
import pandas as pd

df_csv = df_csv.copy()
df_protein = df_protein.copy()

df_csv["protein_id"] = df_csv["protein_id"].astype(str).str.strip()
df_protein["Target_ID"] = df_protein["Target_ID"].astype(str).str.strip()

# normalize sequences (remove whitespace/newlines)
df_csv["sequence"] = df_csv["sequence"].astype(str).str.replace(r"\s+", "", regex=True)
df_protein["Target"] = df_protein["Target"].astype(str).str.replace(r"\s+", "", regex=True)

df_protein = df_protein.rename(columns={"Drug": "SMILES"})

map_df = df_protein[["Target_ID", "Target", "SMILES"]].dropna(subset=["SMILES"])

df_out = df_csv.merge(
    map_df,
    left_on=["protein_id", "sequence"],
    right_on=["Target_ID", "Target"],
    how="left",
    validate="m:m",
).drop(columns=["Target_ID", "Target"])

print("rows before:", len(df_csv))
print("rows after :", len(df_out))
print("missing SMILES rate:", df_out["SMILES"].isna().mean())
```
